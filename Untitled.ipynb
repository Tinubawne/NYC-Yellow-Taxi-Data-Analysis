## mapping payment type

```python
df['payment_type'].value_counts()
```

```python
df.head()
```

```python
payment_mapping = {
    '0': 'Flex Fare',
    '1': 'Credit Card',
    '2': 'Cash',
    '3': 'No Charge',
    '4': 'Dispute'
}

df['payment_type'] = (
    df['payment_type'].astype(str).map(payment_mapping)
)

print(df['payment_type'].value_counts(dropna=False))
```

## mapping RatecodeID

```python
df['RatecodeID'].value_counts()
```

```python
df['RatecodeID'].dtypes
```

```python
RatecodeID_mapping = {
    99.0: 'Unknown',
    1.0: 'Standard rate',
    2.0: 'JFK airport',
    3.0: 'Newark airport',
    4.0: 'Nassau/Westchester',
    5.0: 'Negotiated fare',
    6.0: 'Group ride'
}

df['RatecodeID'] = df['RatecodeID'].map(RatecodeID_mapping)

print(df['RatecodeID'].value_counts(dropna=False))
```
